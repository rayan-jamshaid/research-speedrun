```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df_iii = pd.read_csv('/content/mimic_iii_stats2.csv')

df_iv = pd.read_csv('/content/mimic_iv_stats2.csv')
```

### Step 1: Convert all column names to lowercase

```python
df_iii.columns = df_iii.columns.str.lower()
df_iv.columns = df_iv.columns.str.lower()

print("Columns of df_iii after lowercasing:")
print(df_iii.columns.tolist()[:10]) # Displaying first 10 columns
print("\nColumns of df_iv after lowercasing:")
print(df_iv.columns.tolist()[:10]) # Displaying first 10 columns
```

### Step 2: Extract specified variables

```python
desired_columns = [
    'subject_id', 'age', 'sex', 'insurance', 'myocardial_infarction',
    'congestive_heart_failure', 'peripheral_vascular_disease',
    'cerebrovascular_disease', 'dementia', 'chronic_pulmonary_disease',
    'rheumatic_disease', 'peptic_ulcer_disease', 'diabetes', 'paraplegia',
    'renal_disease', 'malignant_cancer', 'severe_liver_disease', 'aids',
    'mortality_flag', 'rbc_avg', 'wbc_avg', 'sodium_avg', 'anion_gap_avg',
    'creatinine_avg', 'potassium_avg', 'chloride_avg', 'glucose_avg',
    'bun_avg', 'bicarbonate_avg', 'hct_avg', 'hgb_avg', 'plt_avg',
    'rdw_avg', 'calcium_avg', 'ast_avg', 'pt_avg', 'tbil_avg', 'inr_avg',
    'alt_avg', 'alp_avg', 'aptt_avg', 'albumin_avg', 'weight_avg',
    'heart_rate_avg', 'respiratory_rate_avg', 'o2_avg', 'dbp_avg',
    'sbp_avg', 'temperature_avg', 'base_excess_avg', 'bmi', 'height_avg',
    'bmi_avg', 'fibrinogen_avg', 'map_avg', 'dbil_avg'
]

# Filter columns for df_iii
df_iii = df_iii[df_iii.columns.intersection(desired_columns)]

# Filter columns for df_iv
df_iv = df_iv[df_iv.columns.intersection(desired_columns)]

print("Columns of df_iii after extraction:")
print(df_iii.columns.tolist())
print("\nColumns of df_iv after extraction:")
print(df_iv.columns.tolist())
```

### Step 3: Remove '_avg' suffix from column names

```python
df_iii.columns = df_iii.columns.str.replace('_avg', '', regex=False)
df_iv.columns = df_iv.columns.str.replace('_avg', '', regex=False)

print("Columns of df_iii after removing '_avg':")
print(df_iii.columns.tolist())
print("\nColumns of df_iv after removing '_avg':")
print(df_iv.columns.tolist())
```

### Step 4: Compute BMI

Assuming `height` is in cm and `weight` is in kg, BMI is calculated as `weight (kg) / (height (m))^2`.

```python
# Compute BMI for df_iii
if 'weight' in df_iii.columns and 'height' in df_iii.columns:
    df_iii['bmi'] = df_iii['weight'] / ((df_iii['height'] / 100)**2)
    print("BMI computed for df_iii. Displaying first 5 BMI values:")
    display(df_iii['bmi'].head())
else:
    print("Cannot compute BMI for df_iii: 'weight' or 'height' column not found.")

print("\n")

# Compute BMI for df_iv
if 'weight' in df_iv.columns and 'height' in df_iv.columns:
    df_iv['bmi'] = df_iv['weight'] / ((df_iv['height'] / 100)**2)
    print("BMI computed for df_iv. Displaying first 5 BMI values:")
    display(df_iv['bmi'].head())
else:
    print("Cannot compute BMI for df_iv: 'weight' or 'height' column not found.")
```

### Step 5: Insurance column

```python
# Define a function to convert insurance status to binary
def convert_insurance_to_binary(insurance_status):
    if pd.isna(insurance_status):
        return 'No Insurance'
    else:
        # Assuming any non-NaN value means the patient has some form of insurance
        return 'Insurance'

# Process df_iii
print("Processing df_iii 'insurance' column:")
if 'insurance' in df_iii.columns:
    print("Unique values in df_iii['insurance'] before transformation:")
    print(df_iii['insurance'].value_counts(dropna=False))

    # Explicitly create a copy to avoid SettingWithCopyWarning
    df_iii_copy = df_iii.copy()
    df_iii_copy['insurance_binary'] = df_iii_copy['insurance'].apply(convert_insurance_to_binary)
    df_iii = df_iii_copy.drop(columns=['insurance']) # Update df_iii with the processed copy

    print("\nUnique values in df_iii['insurance_binary'] after transformation:")
    print(df_iii['insurance_binary'].value_counts(dropna=False))
else:
    print("Error: 'insurance' column not found in df_iii. Skipping transformation for df_iii.")

print("\n" + "-"*50 + "\n") # Separator for better readability

# Process df_iv
print("Processing df_iv 'insurance' column:")
if 'insurance' in df_iv.columns:
    print("Unique values in df_iv['insurance'] before transformation:")
    print(df_iv['insurance'].value_counts(dropna=False))

    # Explicitly create a copy to avoid SettingWithCopyWarning
    df_iv_copy = df_iv.copy()
    df_iv_copy['insurance_binary'] = df_iv_copy['insurance'].apply(convert_insurance_to_binary)
    df_iv = df_iv_copy.drop(columns=['insurance']) # Update df_iv with the processed copy

    print("\nUnique values in df_iv['insurance_binary'] after transformation:")
    print(df_iv['insurance_binary'].value_counts(dropna=False))
else:
    print("Error: 'insurance' column not found in df_iv. Skipping transformation for df_iv.")
```

```python
# Define the mapping for insurance_binary
insurance_mapping = {'Insurance': 1, 'No Insurance': 0}

# Process df_iii
print("Processing df_iii 'insurance_binary' column:")
if 'insurance_binary' in df_iii.columns:
    print("Unique values in df_iii['insurance_binary'] before mapping:")
    print(df_iii['insurance_binary'].value_counts(dropna=False))

    # Apply the mapping and rename the column
    # Create a copy to avoid SettingWithCopyWarning
    df_iii = df_iii.copy()
    df_iii['insurance_binary'] = df_iii['insurance_binary'].map(insurance_mapping)
    df_iii = df_iii.rename(columns={'insurance_binary': 'insurance'})

    print("\nUnique values in df_iii['insurance'] after mapping:")
    print(df_iii['insurance'].value_counts(dropna=False))
else:
    print("Error: 'insurance_binary' column not found in df_iii. Skipping mapping for df_iii.")

print("\n" + "-"*50 + "\n") # Separator for better readability

# Process df_iv
print("Processing df_iv 'insurance_binary' column:")
if 'insurance_binary' in df_iv.columns:
    print("Unique values in df_iv['insurance_binary'] before mapping:")
    print(df_iv['insurance_binary'].value_counts(dropna=False))

    # Apply the mapping and rename the column
    # Create a copy to avoid SettingWithCopyWarning
    df_iv = df_iv.copy()
    df_iv['insurance_binary'] = df_iv['insurance_binary'].map(insurance_mapping)
    df_iv = df_iv.rename(columns={'insurance_binary': 'insurance'})

    print("\nUnique values in df_iv['insurance'] after mapping:")
    print(df_iv['insurance'].value_counts(dropna=False))
else:
    print("Error: 'insurance_binary' column not found in df_iv. Skipping mapping for df_iv.")
```

### Step 6: Apply Label Encoding to Sex

```python
from sklearn.preprocessing import LabelEncoder

# Initialize LabelEncoder
le = LabelEncoder()

# Process df_iii
print("Processing df_iii 'sex' column:")
if 'sex' in df_iii.columns:
    print("Unique values in df_iii['sex'] before transformation:")
    print(df_iii['sex'].value_counts(dropna=False))

    # Explicitly create a copy to avoid SettingWithCopyWarning
    df_iii_copy = df_iii.copy()
    df_iii_copy['sex'] = le.fit_transform(df_iii_copy['sex'])
    df_iii = df_iii_copy # Update df_iii with the processed copy

    print("\nUnique values in df_iii['sex'] after transformation:")
    print(df_iii['sex'].value_counts(dropna=False))
else:
    print("Error: 'sex' column not found in df_iii. Skipping transformation for df_iii.")

print("\n" + "-"*50 + "\n") # Separator for better readability

# Process df_iv
print("Processing df_iv 'sex' column:")
if 'sex' in df_iv.columns:
    print("Unique values in df_iv['sex'] before transformation:")
    print(df_iv['sex'].value_counts(dropna=False))

    # Explicitly create a copy to avoid SettingWithCopyWarning
    df_iv_copy = df_iv.copy()
    df_iv_copy['sex'] = le.fit_transform(df_iv_copy['sex'])
    df_iv = df_iv_copy # Update df_iv with the processed copy

    print("\nUnique values in df_iv['sex'] after transformation:")
    print(df_iv['sex'].value_counts(dropna=False))
else:
    print("Error: 'sex' column not found in df_iv. Skipping transformation for df_iv.")
```
